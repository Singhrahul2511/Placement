#### 3. Return all divisor of any paticular number

```python
class Solution:
    def divisors(self, n):
        l = []
        for i in range(1, int(abs(n**0.5)) + 1):
            if n % i == 0:
                l.append(i)
                if i != n // i:
                    l.append(n // i)
        return sorted(l)  # ✅ return instead of print

obj = Solution()
print(obj.divisors(36))  # Now it prints: [1, 13]
```

### 4. palindrome check-->reverse krne pr bhi whi aa jaye number whi hai palindrome

```python
class Solution:
    def palindrome(self,n):
        dup =n
        rev= 0
        while n>0:
            last =n%10
            n=n//10
            rev= rev*10+last
        return rev==dup
obj = Solution()
obj.palindrome(1113)
            
```

#### 7. To find the missing number from the given series

```python
def miss(arr,n):
    t_sum = n*(n+1)//2
    a_sum = sum(arr)
    return t_sum - a_sum
arr =[1,2,3,4,6,7]
n =7
miss(arr,n)
```

### 13. Factorial of a number.

```python
class Solution():
    @staticmethod
    def fact(n):
        fact = 1
        for i in range(1,n+1):
            fact *=i
        return fact
obj = Solution()
obj.fact(5)
```

### 14. second largest number

```python
class Solution:
    def getSecondLargest(self, arr):
        # Code Here
        arr.sort()
        largest = arr[-1]
        for i in range(len(arr)-1,-1,-1):
            if arr[i]<largest:
                return arr[i]         
obj = Solution()
arr = [23,4,3,2,2]
obj.getSecondLargest(arr)
```

### 15.move all zeroes to the end

```python
class Solution:
    @staticmethod
    def moveZeroes(nums):

        """
        Do not return anything, modify nums in-place instead.
        """
        i = 0
        for j in range(len(nums)):
            if nums[j] !=0:
                nums[i],nums[j]=nums[j],nums[i]
                i +=1
        return nums
obj = Solution()
nums = [1,2,0,403,0,0]
obj.moveZeroes(nums)
```

### 16. rotate an array when value given

```python
# User function Template for python3
class Solution:
    # Function to rotate an array by d elements in counter-clockwise direction.
    def rotateArr(self, arr, d):
        n = len(arr)
        k = d % n   # Handle cases when d > n
        return arr[k:] + arr[:k]
obj = Solution()
obj.rotateArr([1,2,3,4,5],2)
```
