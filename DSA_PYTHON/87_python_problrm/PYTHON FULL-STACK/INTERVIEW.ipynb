- `4. Diff ways to concatinate 2 tuples`

```python
# 1. Using (+) operator --> most common way
t1 = (1,2)
t2 = (3,4)
res = t1+t2
print(res)
```

```python
# 2. using tuple + list concatenation(Good when modifying elements is required.)

t1 = (1,2)
t2 = (3,4)
res = tuple(list(t1)+list(t2))
print(res)
```

```python
# 3) Using Unpacking (Pythonic & Recommended)

t1 = (1,2)
t2 = (3,4)
res = (*t1,*t2)
print(res)

# Most pythonic
#Works for any number of tuples
```

```python
# 4) Using sum() with empty tuple

t1 = (1,2)
t2 = (3,4)
res = sum((t1,t2),())
print(res)

#best to explain in interview
```

```python
# 5) Using itertools.chain() (Best for performance)
import itertools
t1 = (1,2)
t2 = (3,4)
res = tuple(itertools.chain(t1,t2))
print(res)
```

```python
# 10. Decorator Functions
#Functions that modify the behavior of another function.

def deco(fn):
    def wrapper():
        print("Before")
        fn()
        print("After")
    return wrapper

@deco
def hello():
    print("Hello")

hello()
```

# ************* FULL-STACK PYTHON *************

`Yes, Python is considered a platform-independent programming language. This means that a Python program written on one operating system (like Windows) can typically be executed on another operating system (like macOS or Linux) without requiring modifications to the source code.

This platform independence is achieved through the following mechanism:


Bytecode Compilation: When you run a Python program, the source code is first compiled into an intermediate form called bytecode. This bytecode is not specific to any particular operating system or hardware architecture.


Python Virtual Machine (PVM): Each platform has its own Python interpreter, which includes a Python Virtual Machine (PVM). The PVM is responsible for executing the Python bytecode on that specific platform. It translates the platform-independent bytecode into machine code that the underlying operating system can understand and execute.


Therefore, as long as a Python interpreter is installed on a given platform, the same Python code can be run on it, making Python a highly portable language.`

` 1. Can be dictionary be concatinated`

```python
#Dictionaries cannot be directly concatenated using the + operator in Python, unlike lists or tuples.
# but there is several ways can do it

# 1.Using the update() method: This method adds key-value pairs from one dictionary into another. If keys are duplicated, the values from the updated dictionary will overwrite the values in the original dictionary.

dict1 = {'a':1,'b':2,'c':3}
dict2 = {'a':1,'d':4,'e':4,'f':5} # if duplicate will occure it will remove it
dict1.update(dict2)
print(dict1)
```

```python
#2.Using the ** operator (for Python 3.5+): This unpacks the dictionaries into a new one. If keys are duplicated, the value from the dictionary that appears later in the unpacking will be used.


dict1 = {'a':1,'b':2,'c':3}
dict2 = {'d':4,'e':5,'f':5}
merged_dict = {**dict1,**dict2}
print(merged_dict)

```

```python
#3. Using collections.ChainMap (for Python 3.3+): This creates a single view of multiple dictionaries. It doesn't actually merge them but allows you to look up keys across them. Changes are only made to the first dictionary in the chain.


from collections import ChainMap
dict1 = {'a':1,'b':2,'c': 3}
dict2 = {'d':4,'e':5,'f':6}
chain = ChainMap(dict1,dict2)
print(chain)

print(chain['a']) # using this we can access value of any key
```

` 2. How to iterate through Dictionary`

```python
# 1. Iterating over Keys: This is the default way to iterate if you use a for loop directly on the dictionary. You can also explicitly use the .keys() method.

my_dict = {'name': 'Rahul','age': 24,'city': 'Aurangabad'}
for key in my_dict: # or for key in my_dict.kry():
    print(key)


# 2. Iterating over Values: Use the .values() method to get only the values from the dictionary.
my_dict = {'name': 'Rahul','age': 24,'city': 'Aurangabad'}
for value in my_dict.values():
    print(value)



# 3. Iterating over Key-Value Pairs (Items): Use the .items() method, which returns each pair as a tuple containing the key and its corresponding value. This is often the most useful method when you need both parts.


my_dict = {'name': 'Rahul','age': 24,'city': 'Aurangabad'}

for key,value in my_dict.items():
    print(f"{key} : {value}")
```

` 4. How Memory Management work in python?`

`- Python Memory Manager and Private Heap Space: Python does manage its memory through a dedicated memory manager. This manager allocates memory for Python objects in a private heap space. This means Python controls its own memory area rather than directly relying on the underlying operating system's memory management for every single object.`


`- Objects Not Accessible to the Programmer: The objects within this private heap are indeed managed by Python internally, and programmers typically interact with them through Python's object model and variables, not by directly accessing raw memory addresses.`


`- Garbage Collector: Python includes a garbage collector (specifically, a reference counting system combined with a cyclic garbage collector) that automatically reclaims memory from objects that are no longer referenced or needed, preventing memory leaks.`

`For the question "how to delete that memory" after a = 60, the most accurate answer for Python is that you generally don't directly delete memory yourself like in lower-level languages. Python's automatic memory management handles it.`


- Here's a more detailed correct explanation:

`Removing References:` When you set a = 60, a becomes a reference to the integer object 60. To make that object eligible for memory deallocation, you need to remove all references to it.

`del Statement:` The del statement (del a) removes the variable name a from the current scope. If a was the only reference to the object 60, then the object 60 becomes unreachable.

`Garbage Collection:` Once an object is unreachable (has no references pointing to it), Python's garbage collector will automatically deallocate its memory when it determines it's appropriate. This process is automatic and non-deterministic, meaning you don't control when the memory is freed.

`Reassignment:` If you reassign a (e.g., a = 50), the variable a now refers to the new object 50. The original object 60 no longer has a as a reference. If no other variables refer to 60, it becomes eligible for garbage collection.
So, while del a is a way to make an object eligible for collection, it's Python's garbage collector that actually frees the memory, not the del statement itself. The interviewee was correct in mentioning del a as a way to trigger this process (5:47).
