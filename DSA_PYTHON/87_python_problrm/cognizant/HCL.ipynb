```python
# Gcd using eculidian division method

class Solution:
    def GCD(self,a,b):
        while b !=0:
            a,b = b, a%b
        return a
obj = Solution()
obj.GCD(65,15)
```

```python
## ARMSTRONG NUMBER

class Solution:
    def arms(self,num):
        dup = num
        res = 0
        n= len(str(dup))
        while num>0:
            lst = num%10
            res += lst**n
            num = num//10
        if res == dup:
            return True
        else:
            return False
obj = Solution()
obj.arms(153)
```

```python
# PRIME NUMBER PRINTING BETWEEN 2 NUMBER RANGE

class Solution:
    def primerange(self,x,y):
        res = []
        for n in range(x,y+1):
            if n<1:
                continue
            is_prime =True
            for i in range(2,int(n**0.5)+1):
                if n % i == 0:
                    is_prime = False
                    break
            if is_prime:
                res.append(n)

        print(res if res else "no")
obj = Solution()
obj.primerange(3,11)
```
